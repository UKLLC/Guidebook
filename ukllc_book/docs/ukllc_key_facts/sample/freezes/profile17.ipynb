## Introduction

For each freeze, UK LLC produces a summary of the **demographic characteristics** of LPS participants in the UK LLC TRE, including their **sex**, **gender**, **year of birth** (as a proxy for age) and **ethnic group**. This information is drawn from two sources:  
* [**Harmonised LPS demographic variables**](../../ukllc_managed_data/datasets/sociodemo_harmonised/sociodemo_harmonised.md) created by UK LLC.
* [**Demographic information**](../../linked_health_data/nhs_england/registration_datasets/demographics/demographics.ipynb) provided by NHS England (NHSE).

Further details on how these demographic summaries are generated, is available on the [**UK LLC sample and profile**](https://guidebook.ukllc.ac.uk/docs/ukllc_key_facts/sample/ukllc_sample.html) page.
